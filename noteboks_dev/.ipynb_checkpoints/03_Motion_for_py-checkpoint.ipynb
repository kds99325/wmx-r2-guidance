# 🏃 Notebook 3: Motor Control — Velocity Jogging & Position Move

This notebook sends the first real movement commands to axes that already completed Servo On (Notebook 2). It is split into two parts that mirror the two supported entries of `set_mode`:

* **Part A — Velocity Mode (`1`)**: threaded HMI jogging with a dedicated hardware **E-Stop**.
* **Part B — Position Mode (`0`)**: switch back and send a single absolute move command.

> 🚧 **Torque Mode (`2`)** exists as an enum value in `set_mode`, but the currently installed `wmx_r2_message` package does not yet expose a command topic/service for it. It is intentionally skipped here and left as a future extension point (see `00_Plan.md`).

*Prerequisite: You must execute `01_wmx_system_startup.ipynb` and `02_Axis_Activation&Coordinate_Alignment.ipynb` first so the target axes are Servo-On and homed.*

````python
import rclpy
from wmx_r2_message.msg import AxisVelocity
from wmx_r2_message.srv import SetAxis
import ipywidgets as widgets
from IPython.display import display
import threading
import time
from wmx_utils import WmxClient

if not rclpy.ok():
    rclpy.init()

class WmxHmiController(WmxClient):
    def __init__(self):
        super().__init__(node_name='wmx_hmi_controller')

        self.vel_pub = self.create_publisher(AxisVelocity, '/wmx/axis/velocity', 10)

        self.direction = 0
        self.is_running = False
        self.position = 0.0

        self.lbl_status = widgets.Label(value="Status: IDLE (Stopped)")
        self.lbl_pos = widgets.HTML(value="<h3>Current Position: 0.0 deg</h3>")

    def _loop(self):
        target_speed = 100.0 if self.direction > 0 else -100.0
        speeds = [target_speed] * len(self.axis_list)
        accs = [100.0] * len(self.axis_list)
        decs = [100.0] * len(self.axis_list)
        zero_speeds = [0.0] * len(self.axis_list)

        try:
            while self.is_running:
                # 50ms interval loop (20 FPS) for real-time browser feedback simulation
                self.publish_vel(self.axis_list, speeds, accs, decs)
                self.position += self.direction * 1.5
                self.lbl_pos.value = f"<h3>Current Position: <span style='color: #007acc;'>{self.position:.1f} deg</span></h3>"
                time.sleep(0.05)
        finally:
            # Always publish a zero-velocity stop, even if the loop exits
            # because of an error/exception, not just a normal STOP click.
            # NOTE: this only stops the motor if the browser/kernel is still
            # alive to run this line. It is not a substitute for a hardware
            # or WMX3-side communication watchdog for true E-stop safety.
            self.publish_vel(self.axis_list, zero_speeds, accs, decs)

    def publish_vel(self, index, velocity, acc, dec):
        msg = AxisVelocity()
        msg.index = index
        msg.velocity = velocity
        msg.acc = acc
        msg.dec = dec
        self.vel_pub.publish(msg)

    def start_move(self, direction):
        if not self.is_running:
            self.direction = direction
            self.is_running = True
            dir_text = "Rotating Positive (CW) 🔄" if direction > 0 else "Rotating Negative (CCW) 🔄"
            self.lbl_status.value = f"Status: {dir_text}"
            threading.Thread(target=self._loop, daemon=True).start()

    def stop_move(self):
        self.is_running = False
        self.direction = 0
        self.lbl_status.value = "Status: IDLE (Stopped)"

    def emergency_stop(self):
        """Hard stop via the '/wmx/axis/stop' service, distinct from the graceful
        zero-velocity publish that stop_move() performs. Use this to verify the
        driver-level stop path independently of the jogging loop above."""
        self.is_running = False
        self.direction = 0
        self.lbl_status.value = "Status: 🚨 E-STOP TRIGGERED"

        req = SetAxis.Request()
        req.index = self.axis_list
        req.data = [0] * len(self.axis_list)
        return self.call(SetAxis, '/wmx/axis/stop', req)

# Instantiate the controller
hmi = WmxHmiController()
print(f"✅ HMI Controller Ready. Controlling Axes: {hmi.axis_list}")
````

## Step 9: Switch Axes to Velocity Mode

Notebook 02 left the axes in **Position Mode (`0`)** (see its Step 8). Before the jogging dashboard below can spin the motor by commanding a speed, the axes must be switched to **Velocity Mode (`1`)**.

````python
req = SetAxis.Request()
req.index = hmi.axis_list
req.data = [1] * len(hmi.axis_list)

res = hmi.call(SetAxis, '/wmx/axis/set_mode', req)
if res.success:
    print(f"✅ [SUCCESS] Axes switched to Velocity Mode:\n{res.message}")
else:
    print(f"❌ [ERROR] Failed to switch control mode: {res.message}")
````

## Step 10: Interactive Dashboard (HMI) Control

* This control dashboard uses a multi-threaded Python approach combined with `ipywidgets`.
* Click **[▶ Positive (CW)]** or **[◀ Negative (CCW)]** to spin the actual motor. The thread updates the browser with simulated coordinate feedback in real-time.
* Click **[■ STOP]** to gracefully bring the motor to a halt (publishes a zero-velocity command through the same jog loop).
* Click **[🚨 E-STOP]** to test the safety-critical path: this calls the `/wmx/axis/stop` service directly instead of publishing a velocity, independent of the jogging thread.

````python
# Instantiate control dashboard buttons
btn_cw = widgets.Button(description="▶ Positive (CW)", button_style="success", layout=widgets.Layout(width="140px", height="40px"))
btn_stop = widgets.Button(description="■ STOP", button_style="danger", layout=widgets.Layout(width="90px", height="40px"))
btn_ccw = widgets.Button(description="◀ Negative (CCW)", button_style="warning", layout=widgets.Layout(width="140px", height="40px"))
btn_estop = widgets.Button(description="🚨 E-STOP", button_style="danger", layout=widgets.Layout(width="110px", height="40px"))

# Bind buttons directly to controller state machine functions
btn_cw.on_click(lambda _: hmi.start_move(1))
btn_ccw.on_click(lambda _: hmi.start_move(-1))
btn_stop.on_click(lambda _: hmi.stop_move())
btn_estop.on_click(lambda _: hmi.emergency_stop())

# Arrange the GUI layout nicely and display
controls = widgets.HBox([btn_ccw, btn_stop, btn_cw, btn_estop], layout=widgets.Layout(justify_content="center", margin="10px 0"))
display_box = widgets.VBox([hmi.lbl_status, hmi.lbl_pos], layout=widgets.Layout(align_items="center"))
panel = widgets.VBox([controls, display_box], layout=widgets.Layout(border="1px solid #ccc", padding="15px", width="420px", align_items="center"))

print("📬 Operate the motor and monitor the live coordinate stream below:")
display(panel)
````

## Step 11: Switch Axes back to Position Mode

Before sending a target-position command, switch the axes back to **Position Mode (`0`)** — Velocity Mode has no concept of an absolute target position.

````python
req = SetAxis.Request()
req.index = hmi.axis_list
req.data = [0] * len(hmi.axis_list)

res = hmi.call(SetAxis, '/wmx/axis/set_mode', req)
if res.success:
    print(f"✅ [SUCCESS] Axes switched back to Position Mode:\n{res.message}")
else:
    print(f"❌ [ERROR] Failed to switch control mode: {res.message}")
````

## Step 12: Send an Absolute Position Move Command

With the axes back in Position Mode, send a single move command to a target position.

> ⚠️ **Verify the service name before running this cell.** Unlike `clear_alarm` / `set_on` / `homing` / `set_mode` (confirmed in Notebook 02), the exact `SetAxis`-family service name for an absolute move (e.g. `move`, `move_abs`, `set_position`) depends on the installed `wmx_r2_message` package version. Check it once with:
> ```bash
> ros2 service list | grep /wmx/axis
> ```
> and update the `SERVICE_NAME` string below to match what you find.

````python
SERVICE_NAME = '/wmx/axis/move_abs'  # TODO: confirm against `ros2 service list | grep /wmx/axis`

req = SetAxis.Request()
req.index = hmi.axis_list
req.data = [50, 50][:len(hmi.axis_list)]  # target position per axis, in the unit defined by the loaded XML params

res = hmi.call(SetAxis, SERVICE_NAME, req)
if res.success:
    print(f"✅ [SUCCESS] Move command accepted:\n{res.message}")
else:
    print(f"❌ [ERROR] {res.message}")
````
