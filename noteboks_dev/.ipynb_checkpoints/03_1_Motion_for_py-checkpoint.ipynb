## Step 9: Quick Mode Examples (Position vs Velocity)

Before jumping into the threaded HMI dashboard below, it's worth seeing each control mode
run **once, on its own, with a plain script** — no widgets, no background thread. This uses a
lightweight `WmxClient` connection (separate node name, `wmx_client_03`) so it doesn't collide
with the `wmx_hmi_controller` node created later in this notebook.

Recall the mode table from Notebook 02, Step 8:

| Code | Mode | Behavior |
|---|---|---|
| `0` | Position | Moves to a target and stops automatically on arrival |
| `1` | Velocity | Spins continuously until an explicit stop (velocity `0`) command |

*Prerequisite: axes must already be Servo On (Notebook 02, Steps 5–7).*

```python
import rclpy
from wmx_r2_message.srv import SetAxis
from wmx_r2_message.msg import AxisPose, AxisVelocity
from wmx_utils import WmxClient

if not rclpy.ok():
    rclpy.init()

wmx = WmxClient(node_name='wmx_client_03')
print(f"✅ WMX Client Ready for Notebook 03 examples. Target Axes: {wmx.axis_list}")
```

### Step 9-A: Position Mode Example

Axes should already be in Position mode from Notebook 02's Step 8 (this cell re-sets it anyway,
just to make the example runnable standalone). Sends one absolute move and waits `WAIT_SEC`
seconds — long enough for the move to finish — before printing a confirmation.

Change `TARGET_POSITION` to try a different target.

```python
import time

TARGET_POSITION = 1.0     # 목표 위치 (라디안, 혹은 로드된 XML 파라미터 단위)
MOVE_VELOCITY   = 5.0     # 이동 속도
MOVE_ACC        = 10.0    # 가속도
MOVE_DEC        = 10.0    # 감속도
WAIT_SEC        = 3.0     # 이동 완료를 기다릴 시간(초)

# 안전하게 Position 모드 재확인 (0 = Position)
req = SetAxis.Request()
req.index = wmx.axis_list
req.data = [0] * len(wmx.axis_list)
res = wmx.call(SetAxis, '/wmx/axis/set_mode', req)
print(f"Mode → Position: {res.message}" if res.success else f"❌ {res.message}")

pos_pub = wmx.create_publisher(AxisPose, '/wmx/axis/position', 1)

msg = AxisPose()
msg.index = wmx.axis_list
msg.target = [TARGET_POSITION] * len(wmx.axis_list)
msg.velocity = [MOVE_VELOCITY] * len(wmx.axis_list)
msg.acc = [MOVE_ACC] * len(wmx.axis_list)
msg.dec = [MOVE_DEC] * len(wmx.axis_list)

print(f"🎯 Moving to target position {TARGET_POSITION} ...")
pos_pub.publish(msg)

time.sleep(WAIT_SEC)
print(f"✅ Move command issued and {WAIT_SEC}s elapsed. "
      f"Check /wmx/axis/state (motion_complete field) to confirm arrival.")
```

### Step 9-B: Velocity Mode Example

Velocity mode has no target — it just spins at `RUN_VELOCITY` until told to stop. This example
switches the mode to Velocity, runs for `RUN_SEC` seconds, sends an explicit `0` velocity to
stop, then **switches back to Position mode** at the end so the axes are left in the same state
the HMI dashboard below expects.

```python
import time

RUN_VELOCITY = 50.0   # 목표 속도
RUN_ACC      = 10.0   # 가속도
RUN_DEC      = 10.0   # 감속도
RUN_SEC      = 3.0    # 지정 속도로 실행할 시간(초)

# 1. Velocity 모드로 전환 (1 = Velocity)
req = SetAxis.Request()
req.index = wmx.axis_list
req.data = [1] * len(wmx.axis_list)
res = wmx.call(SetAxis, '/wmx/axis/set_mode', req)
print(f"Mode → Velocity: {res.message}" if res.success else f"❌ {res.message}")

vel_pub = wmx.create_publisher(AxisVelocity, '/wmx/axis/velocity', 1)

# 2. 지정 속도로 RUN_SEC초 동안 실행
msg = AxisVelocity()
msg.index = wmx.axis_list
msg.velocity = [RUN_VELOCITY] * len(wmx.axis_list)
msg.acc = [RUN_ACC] * len(wmx.axis_list)
msg.dec = [RUN_DEC] * len(wmx.axis_list)

print(f"🚀 Running at velocity {RUN_VELOCITY} for {RUN_SEC}s ...")
vel_pub.publish(msg)
time.sleep(RUN_SEC)

# 3. 정지 (속도 0)
msg.velocity = [0.0] * len(wmx.axis_list)
vel_pub.publish(msg)
print("🛑 Stopped.")

# 4. 아래 HMI 대시보드를 위해 다시 Position 모드로 복귀
req = SetAxis.Request()
req.index = wmx.axis_list
req.data = [0] * len(wmx.axis_list)
res = wmx.call(SetAxis, '/wmx/axis/set_mode', req)
print(f"Mode → Position (restored): {res.message}" if res.success else f"❌ {res.message}")
```

```python
import rclpy
from wmx_r2_message.msg import AxisVelocity
import ipywidgets as widgets
from IPython.display import display
import threading
import time
from wmx_utils import WmxClient

if not rclpy.ok():
    rclpy.init()

class WmxHmiController(WmxClient):
    def __init__(self):
        super().__init__(node_name='wmx_hmi_controller')

        self.vel_pub = self.create_publisher(AxisVelocity, '/wmx/axis/velocity', 10)

        self.direction = 0
        self.is_running = False
        self.position = 0.0

        self.lbl_status = widgets.Label(value="Status: IDLE (Stopped)")
        self.lbl_pos = widgets.HTML(value="<h3>Current Position: 0.0 deg</h3>")

    def _loop(self):
        target_speed = 100.0 if self.direction > 0 else -100.0
        speeds = [target_speed] * len(self.axis_list)
        accs = [100.0] * len(self.axis_list)
        decs = [100.0] * len(self.axis_list)
        zero_speeds = [0.0] * len(self.axis_list)

        try:
            while self.is_running:
                # 50ms interval loop (20 FPS) for real-time browser feedback simulation
                self.publish_vel(self.axis_list, speeds, accs, decs)
                self.position += self.direction * 1.5
                self.lbl_pos.value = f"<h3>Current Position: <span style='color: #007acc;'>{self.position:.1f} deg</span></h3>"
                time.sleep(0.05)
        finally:
            # Always publish a zero-velocity stop, even if the loop exits
            # because of an error/exception, not just a normal STOP click.
            # NOTE: this only stops the motor if the browser/kernel is still
            # alive to run this line. It is not a substitute for a hardware
            # or WMX3-side communication watchdog for true E-stop safety.
            self.publish_vel(self.axis_list, zero_speeds, accs, decs)

    def publish_vel(self, index, velocity, acc, dec):
        msg = AxisVelocity()
        msg.index = index
        msg.velocity = velocity
        msg.acc = acc
        msg.dec = dec
        self.vel_pub.publish(msg)

    def start_move(self, direction):
        if not self.is_running:
            self.direction = direction
            self.is_running = True
            dir_text = "Rotating Positive (CW) 🔄" if direction > 0 else "Rotating Negative (CCW) 🔄"
            self.lbl_status.value = f"Status: {dir_text}"
            threading.Thread(target=self._loop, daemon=True).start()

    def stop_move(self):
        self.is_running = False
        self.direction = 0
        self.lbl_status.value = "Status: IDLE (Stopped)"

# Instantiate the controller
hmi = WmxHmiController()
print(f"✅ HMI Controller Ready. Controlling Axes: {hmi.axis_list}")
```

### Step 10: Interactive Dashboard (HMI) Control
* This control dashboard uses a multi-threaded Python approach combined with `ipywidgets`.
* Click **[▶ Positive (CW)]** or **[◀ Negative (CCW)]** to spin the actual motor. The thread updates the browser with simulated coordinate feedback in real-time.
* Click **[■ STOP]** to safely bring the physical motor to a halt.

> ⚠️ This dashboard publishes directly to `/wmx/axis/velocity` without setting the mode itself. If Step 9-B above left the axes in **Position** mode (it restores Position mode at the end), switch back to **Velocity** mode first — reuse the `set_mode` snippet from Step 9-B with `data=1` — or the buttons won't produce motion.

```python
# Instantiate control dashboard buttons
btn_cw = widgets.Button(description="▶ Positive (CW)", button_style="success", layout=widgets.Layout(width="140px", height="40px"))
btn_stop = widgets.Button(description="■ STOP", button_style="danger", layout=widgets.Layout(width="90px", height="40px"))
btn_ccw = widgets.Button(description="◀ Negative (CCW)", button_style="warning", layout=widgets.Layout(width="140px", height="40px"))

# Bind buttons directly to controller state machine functions
btn_cw.on_click(lambda _: hmi.start_move(1))
btn_ccw.on_click(lambda _: hmi.start_move(-1))
btn_stop.on_click(lambda _: hmi.stop_move())

# Arrange the GUI layout nicely and display
controls = widgets.HBox([btn_ccw, btn_stop, btn_cw], layout=widgets.Layout(justify_content="center", margin="10px 0"))
display_box = widgets.VBox([hmi.lbl_status, hmi.lbl_pos], layout=widgets.Layout(align_items="center"))
panel = widgets.VBox([controls, display_box], layout=widgets.Layout(border="1px solid #ccc", padding="15px", width="420px", align_items="center"))

print("📬 Operate the motor and monitor the live coordinate stream below:")
display(panel)
```
