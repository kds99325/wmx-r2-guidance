# 🏃 Notebook 3: Threaded Motion Control & Safe Jogging

This notebook verifies physical or simulated hardware motion control. Since capturing low-latency continuous keypresses inside a web browser is unstable, we will run the native keyboard jogging node in a dedicated terminal.

### 🎯 Practice Objectives
1. **Interactive Jogging**: Move the motor using a low-latency keyboard interface.
2. **Safety Watchdog**: Verify how the hardware automatically halts the moment your key is released.

## 1. KEYBOARD JOGGING
1. Open the Jupyter Terminal:
   👉 Click the Jupyter logo -> Select 'New' at the top-right -> Click 'Terminal'.

2. Copy and run this command to start the interactive jogger:
   >ros2 run wmx_r2_package jog_keyboard_node --ros-args -p axis:="1" -p velocity:=100.0 -p acc:=100.0 -p dec:=100.0

3. Test the Safety Watchdog:
   - Click inside the terminal window to focus.
   - Press and HOLD the `d` key. The motor will spin forward.
   - RELEASE the `d` key. Watch the motor immediately decelerate to a stop!
   - Press and HOLD `a` to spin backward, and release to stop.
   - Exit the jogger safely by pressing `q` inside the terminal.
