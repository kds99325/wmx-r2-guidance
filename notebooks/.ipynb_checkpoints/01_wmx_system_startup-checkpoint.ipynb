# ⚙️ Notebook 1: WMX R2 Infrastructure & Engine Startup

This notebook guides you through the systematic startup process to initialize the **WMX3 Software Motion Engine** and transition the physical EtherCAT hardware into a ready-to-run state (**Servo On**).

### 🎯 Objective & Scope
* **Focus**: Executes the hard-deterministic startup sequence from booting network sockets to applying power to the motor coils.
* **Out of Scope**: This notebook does *not* cover motor jogging, motion inputs, or trajectory path planning (handled in subsequent notebooks).
* **Prerequisite**: It assumes that ROS 2 (Jazzy/Humble), Ubuntu 22.04/24.04, and your colcon workspace are already fully built and configured.

### 🔄 State Machine & Initialization Sequence
To prevent sudden hardware movement or damage, WMX R2 enforces a strict step-by-step state transition:
1. **Step 1: S/W Process Bring-up (Server)**: Launch the background ROS 2 nodes in the terminal to open service communication gates.
2. **Step 2: Create WMX3 Device**: Client requests the server to instantiate the core C++ SDK motion engine device in memory.
3. **Step 3: Start EtherCAT Communication**: Power up the network master to synchronize communication with connected slave drives.
4. **Step 4: Load XML Parameters**: Ingest physical parameters (gear ratios, soft limits, encoder profiles) to the active engine.

```python
import rclpy
from std_srvs.srv import SetBool
from wmx_r2_message.srv import SetEngine, LoadWmxParams, SetAxis
from wmx_utils import WmxClient

if not rclpy.ok():
    rclpy.init()

wmx = WmxClient(node_name='wmx_client_01')
print(f"✅ ROS 2 Jupyter Client Ready. Target Axes: {wmx.axis_list}")
```

## Step 2: Create WMX3 Device Interface
We request the backend `wmx_engine_node` server to call the WMX3 C++ SDK API (`CreateDevice`) under the hood, allocating the motion controller instance in memory.

```python
req = SetEngine.Request()
req.data = True
req.path = '/opt/wmx3/'
req.name = 'my_device'

res = wmx.call(SetEngine, '/wmx/engine/set_device', req)
if res is not None:
    print(f"✅ [SUCCESS] {res.message}")
else:
    print("❌ [ERROR] Failed to create device")
```

## Step 3: Start EtherCAT Communication
Command the engine to start the EtherCAT master network, initiating hard-synchronized cyclic packet exchange with connected motor slave drives (OP State).

```python
req = SetBool.Request()
req.data = True

res = wmx.call(SetBool, '/wmx/engine/set_comm', req)
if res is not None and res.success:
    print(f"✅ [SUCCESS] {res.message}")
else:
    print("❌ [ERROR] Failed to start communication.")
```

## Step 4: Load WMX Parameter XML File
Upload physical configuration parameters (including gear ratios, hardware travel limits, and encoder scaling factors) to the engine memory.

```python
import os

xml_path = os.path.expanduser('~/workspaces/movensys_ws/install/wmx_r2_package/share/wmx_r2_package/config/diffbot_wmx_parameters.xml')

req = LoadWmxParams.Request()
req.file_path = xml_path

res = wmx.call(LoadWmxParams, '/wmx/params/load', req)
if res is not None:
    print(f"✅ XML parameters injected: \n{xml_path}")
else:
    print("❌ [ERROR] Failed to load parameter XML file.")
```
