# 🏁 Notebook 2: Axis Activation & Coordinate Alignment
This notebook guides you through waking up and aligning the controlled axes to prepare the physical actuators for precise motion.

## 🎯 Objective & Scope
* **Focus**: Systematically clears active drive faults, locks the servos with power, establishes absolute coordinate origins, and configures the control modes.
* **Out of Scope**: This notebook does not cover real-time interactive HMI dashboards, manual jogging, or continuous trajectory execution (handled in subsequent notebooks).
* **Prerequisite**: It assumes that Notebook 1 (Infrastructure & Engine Startup) has completed successfully, ensuring the WMX3 core engine and EtherCAT master are fully active.

## 🔄 State Machine & Initialization Sequence
To prevent sudden hardware movement or damage, WMX R2 enforces a strict step-by-step state transition:

* **Step 5: Clear Axis Alarms**: Resolve active hardware alarm states and clear fault latches on the servo drives.
* **Step 6: Motor Servo On**: Enable current injection to the motor coils to lock the axes securely in place.
* **Step 7: Axis Homing (Zero-Definition)**: Execute a physical homing routine to align software coordinates with the physical origin (defines absolute `0.0`, does not spin back).
* **Step 8: Set Control Mode (Position Mode)**: Configure the axes to Position Control Mode (0) to prepare them for safe, interpolated position-based trajectory execution.


```python
import rclpy
from wmx_r2_message.msg import AxisVelocity
from wmx_r2_message.srv import SetAxis
import ipywidgets as widgets
from IPython.display import display
import threading
import time

from wmx_utils import WmxClient

if not rclpy.ok():
    rclpy.init()

wmx = WmxClient(node_name='wmx_client_02')
print(f"✅ WMX Client Ready for Notebook 02. Target Axes: {wmx.axis_list}")
```

## Step 5: Clear Axis Alarms
Reset any error latches or safety faults currently present on the servo drives (e.g., initial startup error, emergency stop release) to prepare them for power injection.

```python
req = SetAxis.Request()
req.index = [0]
req.data = [0]

res = wmx.call(SetAxis, '/wmx/axis/clear_alarm', req)
if res is not None:
    print(f"✅ [SUCCESS]\n{res.message}".replace("; ", "\n").strip())
else:
    print("❌ [ERROR] Failed to request alarm clear.")
```

## Step 6: Motor Servo On
*   Calls the **/wmx/axis/set_on** service to apply power to the physical axis and release the brake, holding the hardware in a ready-to-control state.

```python
req = SetAxis.Request()
req.index = [0]
req.data = [0]

res = wmx.call(SetAxis, '/wmx/axis/set_on', req)
if res is not None:
    print(f"{res.message}".replace("; ", "\n").strip())
else:
    print("❌ [ERROR] Failed to enable servo (Servo On).")
```

## Step 7

```python
req = SetAxis.Request()
req.index = wmx.axis_list
req.data = [0]

res = wmx.call(SetAxis, '/wmx/axis/homing', req)
if res is not None:
    msg = res.message.replace('; ', '\n').replace(';', '\n').replace(', Axis', '\nAxis').replace(',Axis', '\nAxis')
    print(f"✅ [SUCCESS] Homing Results (Zero registry updated):\n{msg}")
else:
    print("❌ [ERROR] Failed to complete homing sequence.")
```

## Step 8: Set Control Mode

```python
req = SetAxis.Request()
req.index = wmx.axis_list
req.data = [0]  

res = wmx.call(SetAxis, '/wmx/axis/set_mode', req)
if res is not None:
    msg = res.message.replace('; ', '\n').replace(';', '\n').replace(', Axis', '\nAxis').replace(',Axis', '\nAxis')
    print(f"🚀 [SUCCESS] Control Mode Results:\n{msg}")
    print("\nAll systems are fully homed and locked in Position Mode! Ready to move.")
else:
    print("❌ [ERROR] Failed to configure control mode.")
```
